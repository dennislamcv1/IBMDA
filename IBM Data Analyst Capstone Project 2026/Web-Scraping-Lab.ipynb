## Extract information from the given web site
You will extract the data from the below web site: <br> 


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
# Import required libraries for web scraping and data handling
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

Download the webpage at the url


```python
# Send a GET request to the URL and store the response
response = requests.get(url)
print("Status code:", response.status_code)
```

Create a soup object


```python
# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')
print(soup.prettify()[:500])  # Preview the first 500 characters of parsed HTML
```

Scrape the `Language name` and `annual average salary`.


```python
# Find the table in the HTML page
table = soup.find('table')

# Extract all rows from the table body (skip the header row)
rows = table.find_all('tr')

languages = []
salaries  = []

# Iterate over rows, skipping the header (index 0)
for row in rows[1:]:
    cols = row.find_all('td')
    if len(cols) >= 3:                          # columns: No., Language, Avg Salary
        language = cols[1].get_text(strip=True) # 2nd column → Language name
        salary   = cols[2].get_text(strip=True) # 3rd column → Average annual salary
        languages.append(language)
        salaries.append(salary)

# Display the scraped data
for lang, sal in zip(languages, salaries):
    print(f"{lang}: {sal}")
```

Save the scrapped data into a file named *popular-languages.csv*


```python
# Create a DataFrame from the scraped lists
df = pd.DataFrame({
    'Language': languages,
    'Average Annual Salary': salaries
})

# Save to CSV file
df.to_csv('popular-languages.csv', index=False)

print("Data saved to 'popular-languages.csv'")
print(df)
```
